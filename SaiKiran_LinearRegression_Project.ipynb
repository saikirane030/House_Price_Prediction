```python
### I imported the libraries required for data analysis, visualization and linear regression.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
```

```python
### I loaded the real-world housing dataset to explore and build a linear regression model.
```

```python
df = pd.read_csv('housing_data.csv')
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
### I selected above-ground living area as the input feature and SalePrice as the target.
```

```python
X = df[['Gr Liv Area']]
y = df['SalePrice']

print(X.shape)
print(y.shape)
```

```python
### I split the data into training and testing sets before further data preparation.
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

print(f"X_train --> {X_train.shape}")
print(f"X_test --> {X_test.shape}")
print(f"y_train --> {y_train.shape}")
print(f"y_test --> {y_test.shape}")
```

```python
### I checked the selected feature and target for missing values before training the model.
```

```python
print(X_train.isnull().sum())
print(y_train.isnull().sum())
```

```python
### I checked for duplicate rows to make sure repeated records do not affect the analysis
```

```python
df.duplicated().sum()
```

```python
X_train.describe().round().T
```

```python
## building  a model
lr_model = LinearRegression()
lr_model
```

```python
## I trained the linear regression model using the training dataset.
```

```python
lr_model.fit(X_train, y_train)
```

```python
lr_model.coef_
```

```python
#Intercept
```

```python
lr_model.intercept_
```

```python
y_pred_train = lr_model.predict(X_train)
y_pred_test = lr_model.predict(X_test)

print(y_pred_train.shape)
print(y_pred_test.shape)
```

```python
## MSE
```

```python
print(f"Training MSE --> {mean_squared_error(y_train,y_pred_train):.0f}")
print(f"Training RMSE --> {root_mean_squared_error(y_train,y_pred_train):.0f}")
print(f"Training R-squared --> {r2_score(y_train,y_pred_train):.4f}")

print(f"Testing MSE --> {mean_squared_error(y_test,y_pred_test):.0f}")
print(f"Testing RMSE --> {root_mean_squared_error(y_test,y_pred_test):.0f}")
print(f"Testing R-squared --> {r2_score(y_test,y_pred_test):.4f}")
```

```python
import joblib

saved_package = {
    "model": lr_model,
    "features": X.columns.tolist(),
    "metrics": {
        "test_mae": mean_absolute_error(y_test, y_pred_test),
        "test_rmse": root_mean_squared_error(y_test, y_pred_test),
        "test_mse": mean_squared_error(y_test, y_pred_test),
        "test_r2": r2_score(y_test, y_pred_test),
    },
}

joblib.dump(saved_package, "house_model.pkl")
print("Model package saved to house_model.pkl successfully!")
```
